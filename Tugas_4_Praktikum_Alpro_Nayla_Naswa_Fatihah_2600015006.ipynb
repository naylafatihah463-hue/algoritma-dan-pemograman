```python
status_code = 404

if status_code == 200:
  print("OK: Permintaan berhasil.")
elif status_code == 400:
  print("Bad Request: Permintaan tidak valid.")
elif status_code == 404:
  print("Not Found: Halaman tidak ditemukan.")
elif status_code == 500:
  print("Internal Server Error: Kesalahan server.")
else:
  print("Status Code tidak dikenal.")
```

```python
# Contoh Logika setara menggunakan match-case
status_code = 404

match status_code:
    case 200:
        print("OK: Permintaan berhasil.")
    case 400:
        print("Bad Request: Permintaan tidak valid.")
    case 404:
        print("Not Found: Halaman tidak ditemukan.")
    case 500:
        print("Internal Server Error: Kesalahan server.")
    case _:  # Wildcard '_' berfungsi sebagai default / else
        print("Status Code tidak dikenal.")
```

```python
def cek_kategori_hari(hari):
    match hari.title():
      case "Senin" | "Selasa" | "Rabu" | "Kamis" | "Jumat":
        return f"{hari} adalah Hari Kerja (Weekday)."
      case "Sabtu" | "Minggu" :
          return f"{hari} adalah Akhir Pekan (Weekend)."
      case _:
        return "Nama hari tidak valid!"

# Uji Coba Fungsi
print(cek_kategori_hari("senin"))
print(cek_kategori_hari("Sabtu"))
print(cek_kategori_hari ("Kiamat"))
```

```python
def analisis_angka(angka):
    match angka:
        case x if x > 0 and x % 2 == 0:
            print(f"{angka} adalah Bilangan Positif Genap.")
        case x if x > 0 and x % 2 != 0:
            print(f"{angka} adalah Bilangan Positif Ganjil.")
        case x if x < 0:
            print(f"{angka} adalah Bilangan Negatif.")
        case 0:
            print("Angka adalah Nol.")
        case _:
            print("Input bukan angka valid.")

# Uji Coba
analisis_angka(14)
analisis_angka(7)
analisis_angka(-5)
analisis_angka(0)
```

```python
def eksekusi_perintah(command):
    match command:
        case ["move", arah]:
            print(f"Karakter bergerak ke arah: {arah}")
        case ["attack", senjata, target]:
            print(f"Menyerang {target} menggunakan {senjata}!")
        case ["quit"] | ["exit"]:
            print("Keluar dari permainan...")
        case _:
            print("Perintah tidak dikenal!")

# Uji Coba Perintah Game
eksekusi_perintah(["move", "Utara"])
eksekusi_perintah(["attack", "Pedang", "Goblin"])
eksekusi_perintah(["exit"])
```

```python
def proses_pengguna(user_data):
    match user_data:
        case {"nama": nama, "role": "admin"}:
            print(f"Selamat datang Admin {nama}. Akses Penuh Diberikan.")
        case {"nama": nama, "role": "user", "aktif": True}:
            print(f"Halo Pengguna {nama}, akun Anda aktif.")
        case {"nama": nama, "role": "user", "aktif": False}:
            print(f"Peringatan: Akun {nama} dalam status non-aktif!")
        case _:
            print("Format data pengguna tidak valid!")

# Uji coba
proses_pengguna({"nama": "Budi", "role": "admin"})
proses_pengguna({"nama": "Siti", "role": "user", "aktif": True})
proses_pengguna({"nama": "Andi", "role": "user", "aktif": False})
```

```python
# Lengkapi kode fungsi di bawah ini
def hitung_diskon(total_belanja, hari):
    match hari.title():
        case "Jumat" | "Sabtu" if total_belanja >= 100000:
            diskon = 0.15
        case "Jumat" | "Sabtu":
            diskon = 0.05
        case "Minggu":
            diskon = 0.10
        case _:
            diskon = 0.0

    potongan = total_belanja * diskon
    total_bayar = total_belanja - potongan
    return total_bayar

# Uji coba kasir
print("Total bayar (Jumat, 150rb):", hitung_diskon(150000, "jumat"))
print("Total bayar (Minggu, 50rb):", hitung_diskon(50000, "minggu"))
print("Total bayar (Senin, 200rb):", hitung_diskon(200000, "senin"))
```

```python
def periksa_respons_cuaca(data):
    match data:
        # Aturan 1: status 200 dengan kota dan suhu
        case {"status": 200, "kota": kota, "suhu": suhu}:
            print(f"Cuaca di {kota} saat ini adalah {suhu}°C.")

        # Aturan 2: status 404 dengan pesan
        case {"status": 404, "pesan": pesan}:
            print(f"Gagal mendapatkan data: {pesan}")

        # Aturan 3: status 500
        case {"status": 500}:
            print("Layanan API Cuaca sedang gangguan/down.")

        # Aturan 4: Format lainnya
        case _:
            print("Format respons tidak dikenal.")


# Contoh Pengujian:
periksa_respons_cuaca({"status": 200, "kota": "Jakarta", "suhu": 30})
periksa_respons_cuaca({"status": 404, "pesan": "Kota tidak ditemukan"})
periksa_respons_cuaca({"status": 500})
periksa_respons_cuaca({"status": 403, "error": "Forbidden"})
```

```python
def evaluasi_mahasiswa(data):
    match data:
        # Aturan 1: Nilai >= 85 (Guard clause: 0 <= nilai_akhir <= 100)
        case (nama, nilai_akhir) if isinstance(nilai_akhir, (int, float)) and 85 <= nilai_akhir <= 100:
            print(f"{nama} LULUS dengan Predikat A (Sangat Memuaskan)")

        # Aturan 2: Nilai >= 70 dan < 85
        case (nama, nilai_akhir) if isinstance(nilai_akhir, (int, float)) and 70 <= nilai_akhir < 85:
            print(f"{nama} LULUS dengan Predikat B (Memuaskan)")

        # Aturan 3: Nilai >= 60 dan < 70
        case (nama, nilai_akhir) if isinstance(nilai_akhir, (int, float)) and 60 <= nilai_akhir < 70:
            print(f"{nama} LULUS dengan Predikat C (Cukup)")

        # Aturan 4: Nilai < 60 (tetap valid dalam rentang 0-59.9)
        case (nama, nilai_akhir) if isinstance(nilai_akhir, (int, float)) and 0 <= nilai_akhir < 60:
            print(f"{nama} TIDAK LULUS (Perlu Remidial)")

        # Aturan 5: Jika format tuple salah atau nilai di luar rentang 0-100
        case _:
            print("Data mahasiswa tidak valid!")


# Contoh Pengujian:
evaluasi_mahasiswa(("Budi", 88))       # Budi LULUS dengan Predikat A (Sangat Memuaskan)
evaluasi_mahasiswa(("Siti", 75))       # Siti LULUS dengan Predikat B (Memuaskan)
evaluasi_mahasiswa(("Andi", 65))       # Andi LULUS dengan Predikat C (Cukup)
evaluasi_mahasiswa(("Joko", 45))       # Joko TIDAK LULUS (Perlu Remidial)
evaluasi_mahasiswa(("Dewi", 105))      # Data mahasiswa tidak valid! (di luar rentang 0-100)
evaluasi_mahasiswa("BukanTuple")       # Data mahasiswa tidak valid! (format salah)
```
